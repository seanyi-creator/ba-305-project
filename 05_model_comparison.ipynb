# Team A7 - Model comparison
**Question:** Can the details on a San Francisco building permit application predict whether the project will be abandoned rather than completed?

**Context:** Of the 62,148 resolved San Francisco building permits filed between 2000 and 2013, about 36% were never completed (expired, cancelled or withdrawn). Abandoned projects waste DBI review and inspection time. If the risk of abandonment can be estimated at filing, DBI can prioritize high-risk permits for early outreach.

**Purpose:** compare the tuned models from `02` (logistic regression), `03` (decision tree, random forest) and `04` (neural network) on the same data, to choose the model that best fits DBI's goal.

**Methodology:**
1. Refit each model with its tuned settings on the same training set (from `permits_data.py`).
2. Compare ranking ability (cross-validated and test ROC AUC), with bootstrap intervals for the differences.
3. Compare accuracy, recall and precision at the 0.5 cutoff and for the riskiest k% of permits.
4. Compare feature importance and performance on later permits (2011–2013).
5. Turn the best model's predictions into an outreach decision, and compare it with policies that use no model.
6. Diagnose the best model's errors: which permits it gets wrong, and why.

**Target:**
* `y = 1` if the permit did not finish (expired, cancelled, withdrawn, disapproved, revoked or denied)
* `y = 0` if completed
* Recall and precision therefore refer to failing permits.

```python
# --- setup ---
import time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_validate, cross_val_predict
from sklearn.inspection import permutation_importance
from sklearn.exceptions import ConvergenceWarning
from sklearn.metrics import (roc_auc_score, accuracy_score, precision_score, recall_score, f1_score,
                             RocCurveDisplay, ConfusionMatrixDisplay)
warnings.filterwarnings('ignore', category=ConvergenceWarning)
pd.set_option('display.max_columns', 50)
```

## Shared data (from `permits_data.py`)

```python
# --- shared data: cleaning code lives in permits_data.py ---
# In Colab the file isn't there yet, so download it from the team GitHub repo first.
import os, urllib.request
if not os.path.exists('permits_data.py'):
    urllib.request.urlretrieve(
        'https://raw.githubusercontent.com/seanyi-creator/ba-305-project/main/permits_data.py',
        'permits_data.py')

from permits_data import get_permit_data, make_cv, make_preprocessor, RANDOM_STATE
```

```python
# --- load the shared split ---
X_train, X_test, y_train, y_test, groups_train, preprocessor = get_permit_data()
cv = make_cv()
print(f'Train: {len(X_train):,} permits · Test: {len(X_test):,} permits')
print(f'Did not finish: {y_train.mean():.1%} of train, {y_test.mean():.1%} of test')
```

## Tuned settings
| Model | Settings | Tuned in |
|---|---|---|
| Logistic regression | `C = 0.1`, `class_weight = 'balanced'` | `02`, Step 1 |
| Decision tree | `entropy`, `max_depth = 15`, `min_samples_leaf = 200` | `03`, Step 3 |
| Random forest | 200 trees, `entropy`, `max_features = 0.25`, `min_samples_leaf = 5` | `03`, Step 5 |
| Neural network | one hidden layer of 64, `alpha = 0.1`, learning rate 0.001, batches of 256, early stopping | `04`, Steps 5–6 |

```python
# --- the four tuned models ---
COLORS = {'Logistic regression': '#3b6ea5', 'Decision tree': '#7a9a3a',
          'Random forest': '#c0581a', 'Neural network': '#7b4fa0'}

def make_models():
    return {
        'Logistic regression': LogisticRegression(max_iter=3000, C=0.1, class_weight='balanced'),
        'Decision tree': DecisionTreeClassifier(criterion='entropy', max_depth=15, min_samples_leaf=200,
                                                random_state=RANDOM_STATE),
        'Random forest': RandomForestClassifier(n_estimators=200, criterion='entropy', max_features=0.25,
                                                min_samples_leaf=5, random_state=RANDOM_STATE, n_jobs=-1),
        'Neural network': MLPClassifier(hidden_layer_sizes=(64,), alpha=0.1, learning_rate_init=0.001,
                                        batch_size=256, early_stopping=True, n_iter_no_change=10,
                                        max_iter=300, random_state=RANDOM_STATE),
    }

def pipe(model):
    return Pipeline([('prep', make_preprocessor()), ('model', model)])
```

## Step 1 — Model evaluation

```python
# --- Step 1 — cross-validated AUC on the training set, then fit once and score on the test set ---
def scores(y_true, prob, pred):
    return {'test ROC AUC': roc_auc_score(y_true, prob),
            'accuracy': accuracy_score(y_true, pred),
            'recall': recall_score(y_true, pred, zero_division=0),
            'precision': precision_score(y_true, pred, zero_division=0),
            'F1': f1_score(y_true, pred, zero_division=0),
            'share flagged': pred.mean()}

rows, fitted, probs = [], {}, {}
for name, strategy in [('Baseline: always "completed"', 'most_frequent'), ('Baseline: random guess', 'stratified')]:
    d = DummyClassifier(strategy=strategy, random_state=RANDOM_STATE).fit(X_train, y_train)
    rows.append({'model': name, **scores(y_test, d.predict_proba(X_test)[:, 1], d.predict(X_test))})

for name, model in make_models().items():
    cv_auc = cross_validate(pipe(model), X_train, y_train, groups=groups_train, cv=cv,
                            scoring='roc_auc', n_jobs=-1)['test_score'].mean()
    t0 = time.time(); fitted[name] = pipe(model).fit(X_train, y_train); secs = time.time() - t0
    probs[name] = fitted[name].predict_proba(X_test)[:, 1]
    rows.append({'model': name, 'CV ROC AUC': cv_auc,
                 **scores(y_test, probs[name], (probs[name] >= 0.5).astype(int)), 'seconds to fit': secs})

results = pd.DataFrame(rows).set_index('model')
results = results[['CV ROC AUC', 'test ROC AUC', 'accuracy', 'recall', 'precision', 'F1', 'share flagged', 'seconds to fit']]
results.round(3)
```

**Reading the table:**
- **Ranking (test AUC):** random forest **0.698** > neural network 0.659 > decision tree 0.644 > logistic regression 0.633 > baselines 0.50.
- **CV and test agree** for every model (within 0.01), so none of the tuning overfit the training set.
- **Accuracy** ranges from 58.5% to 66.9%, against 63.3% for always predicting "completed". Only the forest clearly beats it.
- **Logistic regression has the highest recall (60.5%)** because its `balanced` weighting flags 49% of permits, not because it ranks better. The other three flag about 18–21%.
- **Speed:** the forest is the slowest to fit (about 15–35 s depending on the machine). The others take seconds.

## Step 2 — ROC curves

```python
fig, ax = plt.subplots(figsize=(6.5, 6))
for name, p in probs.items():
    RocCurveDisplay.from_predictions(y_test, p, name=name, ax=ax, color=COLORS[name])
ax.plot([0, 1], [0, 1], 'k--', lw=1, label='random guessing (AUC = 0.5)')
ax.set_xlabel('false positive rate (completed permits flagged)')
ax.set_ylabel('true positive rate (failing permits caught)')
ax.set_title('Test-set ROC curves'); ax.legend(loc='lower right'); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

**Reading the curves:** the random forest's curve sits above the others across almost the whole range, so it catches more failing permits at any false-alarm rate. The other three curves are close together.

## Step 3 — AUC differences

```python
# --- Step 3 — bootstrap 95% intervals for every pair of models ---
names = list(probs)
pairs = [(a, b) for i, a in enumerate(names) for b in names[i + 1:]]
rng = np.random.default_rng(RANDOM_STATE)
yt = y_test.to_numpy()
diffs = {f'{b} − {a}': [] for a, b in pairs}
for _ in range(1000):
    i = rng.integers(0, len(yt), len(yt))
    auc = {m: roc_auc_score(yt[i], probs[m][i]) for m in names}
    for a, b in pairs:
        diffs[f'{b} − {a}'].append(auc[b] - auc[a])
boot = pd.DataFrame(diffs).quantile([0.025, 0.5, 0.975]).T.set_axis(['2.5%', 'difference', '97.5%'], axis=1)
boot['real difference?'] = np.where((boot['2.5%'] > 0) | (boot['97.5%'] < 0), 'yes', 'no')
boot.round(3)
```

**Reading the intervals:** every interval excludes 0, so every gap between models is real rather than luck.
- The forest beats logistic regression by **0.064**, the tree by 0.053 and the network by 0.039.
- The network beats logistic regression by 0.026 and the tree by 0.014.
- The tree beats logistic regression by only 0.011 (interval 0.002 to 0.020).

*(The bootstrap resamples permits rather than project batches, so the intervals are slightly too narrow. The gaps are large enough that this doesn't change any conclusion.)*

## Step 4 — Recall and precision

```python
metrics = ['accuracy', 'recall', 'precision', 'F1']
model_rows = results.loc[list(probs), metrics]
ax = model_rows.T.plot.bar(figsize=(9, 4), color=[COLORS[m] for m in model_rows.index], rot=0, width=0.8)
ax.axhline(results.loc['Baseline: always "completed"', 'accuracy'], color='grey', ls='--', lw=1,
           label='accuracy of always "completed"')
ax.set_ylim(0, 1); ax.set_ylabel('test set, 0.5 cutoff'); ax.legend(ncol=3, fontsize=8, loc='upper left')
ax.set_title('Accuracy, recall and precision at the default cutoff')
plt.tight_layout(); plt.show()
```

**Reading the chart:** at the 0.5 cutoff the models trade recall for precision. Logistic regression catches the most failures (recall 61%) but only 45% of its flags are correct. The forest has the highest precision (59%) and accuracy (67%). Because the cutoff changes these numbers, the models are ranked on AUC and on the risk-ranking table below.

## Step 5 — Risk ranking

```python
# --- Step 5 — if staff review the riskiest k% of permits, how many really fail? ---
rows = []
for k in [0.05, 0.10, 0.20, 0.30]:
    for name, p in probs.items():
        flag = (p >= np.quantile(p, 1 - k)).astype(int)
        rows.append({'review the riskiest': f'{k:.0%}', 'model': name,
                     'precision': precision_score(y_test, flag), 'recall': recall_score(y_test, flag)})
ranking = pd.DataFrame(rows)
print(f'Base rate (share of test permits that did not finish): {y_test.mean():.1%}')
(ranking.pivot(index='review the riskiest', columns='model', values=['precision', 'recall'])
        .loc[['5%', '10%', '20%', '30%']].round(3))
```

**Reading the table:** with the same review list size, the forest's list is the most accurate at every size. Reviewing the riskiest 10%, **65%** of the forest's picks fail, against 59% for the network and 55% for the tree and logistic regression (base rate 37%). The forest also catches the most failures at every size (18% of all failures in the top 10%).

## Step 6 — Feature importance

```python
# --- Step 6 — permutation importance (drop in test AUC when a feature is shuffled) ---
imp = {}
for name, m in fitted.items():
    r = permutation_importance(m, X_test, y_test, scoring='roc_auc', n_repeats=5,
                               random_state=RANDOM_STATE, n_jobs=-1)
    imp[name] = pd.Series(r.importances_mean, index=X_test.columns)
imp = pd.DataFrame(imp)
imp = imp.loc[imp.mean(axis=1).sort_values(ascending=False).index]

top = imp.head(10).iloc[::-1]
fig, ax = plt.subplots(figsize=(9, 6))
y, h = np.arange(len(top)), 0.2
for j, name in enumerate(imp.columns):
    ax.barh(y + (1.5 - j) * h, top[name], height=h, color=COLORS[name], label=name)
ax.set_yticks(y); ax.set_yticklabels(top.index)
ax.set_xlabel('drop in test AUC when the feature is shuffled'); ax.legend()
ax.set_title('Top 10 features, all four models')
plt.tight_layout(); plt.show()
```

```python
# rank of each feature within each model (1 = most important)
imp.rank(ascending=False).astype(int).head(10)
```

**Reading the chart:** all four models put **`filing_year` first**, and estimated cost and proposed stories are in the top five for almost all of them. Site permits, plan sets and building use come next. Four models built in different ways agreeing on what matters makes the finding more trustworthy, but it also means every model leans on filing year.

## Step 7 — Validation over time

```python
# --- Step 7 — train on 2000–2010, test on 2011–2013 ---
X_all, y_all = pd.concat([X_train, X_test]), pd.concat([y_train, y_test])
early = (X_all['filing_year'] <= 2010).to_numpy()
print(f'Train 2000–2010: {early.sum():,} permits, {y_all[early].mean():.1%} did not finish')
print(f'Test 2011–2013: {(~early).sum():,} permits, {y_all[~early].mean():.1%} did not finish')

rows = []
for name, model in make_models().items():
    m = pipe(model).fit(X_all[early], y_all[early])
    rows.append({'model': name, 'test AUC, random split (Step 1)': results.loc[name, 'test ROC AUC'],
                 'test AUC, 2011–2013 permits': roc_auc_score(y_all[~early], m.predict_proba(X_all[~early])[:, 1])})
temporal = pd.DataFrame(rows).set_index('model')
temporal['change'] = temporal.iloc[:, 1] - temporal.iloc[:, 0]
temporal.round(3)
```

**Reading the table:** every model loses accuracy on permits filed after its training years, and the failure rate itself falls from 37.6% to 19.6%. The forest drops the most (0.698 → 0.614) and ties with the network at 0.614. Logistic regression and the tree fall to about 0.60. On later permits the models are much closer than on the random split, so any deployment would need retraining on recent permits.

## Step 8 — Decision policy
The random forest is used to decide which permits get outreach. Each kind of error is given a rough cost:
- **False alarm:** outreach to a permit that would have finished anyway. Cost = 1 unit (staff time).
- **Missed failure:** a permit is abandoned with no outreach. Cost = 1, 2, 3 or 5 units (wasted review and inspection time, an unfinished site). The true cost is unknown, so a range is shown.

For each cost ratio the cutoff is chosen on out-of-fold predictions on the training set, then applied once to the test set. The model is compared with two policies that use no model (act on no one, act on everyone) and with a fixed-capacity policy (outreach to the riskiest 10%).

```python
# --- Step 8 — out-of-fold forest predictions on the training set, used only to choose cutoffs ---
BEST = 'Random forest'
oof = cross_val_predict(pipe(make_models()[BEST]), X_train, y_train, groups=groups_train, cv=cv,
                        method='predict_proba')[:, 1]

def cost_per_1000(y, flag, miss_cost, alarm_cost=1):
    missed = ((y == 1) & (flag == 0)).sum()
    false_alarms = ((y == 0) & (flag == 1)).sum()
    return 1000 * (miss_cost * missed + alarm_cost * false_alarms) / len(y)

ytr, yte, p = y_train.to_numpy(), y_test.to_numpy(), probs[BEST]
cutoffs = np.round(np.arange(0.02, 0.99, 0.01), 2)
top10 = (p >= np.quantile(p, 0.9)).astype(int)

rows = []
for ratio in [1, 2, 3, 5]:
    train_cost = [cost_per_1000(ytr, (oof >= c).astype(int), ratio) for c in cutoffs]
    cut = cutoffs[int(np.argmin(train_cost))]
    flag = (p >= cut).astype(int)
    rows.append({'missed failure : false alarm': f'{ratio} : 1',
                 'cutoff (from training CV)': cut,
                 'share flagged': flag.mean(),
                 'act on no one': cost_per_1000(yte, np.zeros_like(yte), ratio),
                 'act on everyone': cost_per_1000(yte, np.ones_like(yte), ratio),
                 'model, riskiest 10%': cost_per_1000(yte, top10, ratio),
                 'model, cost-based cutoff': cost_per_1000(yte, flag, ratio)})
policy = pd.DataFrame(rows).set_index('missed failure : false alarm')
policy['saving vs. best no-model policy'] = (policy[['act on no one', 'act on everyone']].min(axis=1)
                                            - policy['model, cost-based cutoff'])
print('Test-set cost per 1,000 permits (lower is better)')
policy.round(2)
```

```python
# --- Step 8 — cost per 1,000 permits under each policy ---
cols = ['act on no one', 'act on everyone', 'model, riskiest 10%', 'model, cost-based cutoff']
ax = policy[cols].plot.bar(figsize=(9, 4.2), rot=0, width=0.8,
                           color=['#bbbbbb', '#888888', '#e0a27a', '#c0581a'])
ax.set_ylabel('cost per 1,000 permits (test set)'); ax.set_xlabel('cost of a missed failure : cost of a false alarm')
ax.set_title('Random forest vs. policies that use no model'); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```

```python
# --- Step 8 — confusion matrices: default 0.5 vs. the chosen cutoff ---
MAIN_RATIO = '2 : 1'
CUT = policy.loc[MAIN_RATIO, 'cutoff (from training CV)']
fig, axes = plt.subplots(1, 2, figsize=(11, 4.3))
for ax, cut in zip(axes, [0.5, CUT]):
    ConfusionMatrixDisplay.from_predictions(y_test, (p >= cut).astype(int),
                                            display_labels=['completed (0)', 'did not finish (1)'],
                                            cmap='Blues', colorbar=False, ax=ax)
    ax.set_title(f'Random forest, cutoff {cut:.2f}')
plt.tight_layout(); plt.show()
```

**Reading the table and chart:**
- **The model beats both no-model policies at every cost ratio.** At 2 : 1 the best policy without a model is outreach to everyone (633 per 1,000 permits). The forest with a cutoff of 0.34 costs 480, about **24% less**.
- **The best cutoff falls as missed failures get more expensive:** 0.48 at 1 : 1, 0.34 at 2 : 1, 0.25 at 3 : 1 and 0.17 at 5 : 1. At 5 : 1 the model flags 86% of permits and is close to acting on everyone.
- **A fixed riskiest-10% list** beats acting on no one, but not acting on everyone once a miss costs 2× a false alarm or more, because it leaves most failures uncontacted. It is the right policy only when staff capacity is the binding limit.

**Chosen cutoff: 0.34 (2 : 1).** A missed abandonment wastes plan review and inspections and leaves an unfinished site, which plausibly costs more than one outreach contact, but nothing supports a ratio far above 2. At 0.34 the forest flags 54% of permits and catches 72% of failures (against 34% at the default 0.5), and 49% of its flags are real failures. If DBI cannot contact half of new permits, a riskiest-k% list sized to its capacity should be used instead.

## Step 9 — Error analysis
The forest's errors at the chosen cutoff (Step 8, 2 : 1 cost ratio), in three parts: sanity-check a few cases, compare the permits it gets wrong with the ones it gets right, and decide what to do about it.

```python
# --- Step 9 — label every test permit by outcome ---
errs = X_test.copy()
errs['cost ($)'] = (10 ** errs['log_estimated_cost']).round(-2)
errs['actual'] = y_test.map({0: 'completed', 1: 'did not finish'})
errs['predicted risk'] = p.round(3)
flag = p >= CUT
errs['result'] = np.select([(yte == 1) & flag, (yte == 1) & ~flag, (yte == 0) & flag],
                           ['caught failure', 'missed failure', 'false alarm'], 'correct "completed"')
errs['result'].value_counts().to_frame('test permits')
```

### a. Sanity check: a few cases

```python
# --- Step 9a — the riskiest, the safest, a typical permit, the most confident miss and false alarm ---
show = ['filing_year', 'permit_type_definition', 'cost ($)', 'cost_is_placeholder', 'plansets',
        'number_of_proposed_stories', 'proposed_use', 'site_permit', 'predicted risk', 'actual', 'result']
cases = {
    'riskiest permit': errs['predicted risk'].idxmax(),
    'safest permit': errs['predicted risk'].idxmin(),
    'typical permit (median risk)': (errs['predicted risk'] - errs['predicted risk'].median()).abs().idxmin(),
    'most confident miss': errs.loc[errs['result'] == 'missed failure', 'predicted risk'].idxmin(),
    'most confident false alarm': errs.loc[errs['result'] == 'false alarm', 'predicted risk'].idxmax(),
}
errs.loc[list(cases.values()), show].set_axis(list(cases), axis=0).T
```

**Reading the cases:**
- **Riskiest (risk 0.99, did not finish):** a $9,600 alteration to an apartment building filed in 2004. Older, cheap alterations are the highest-risk group.
- **Safest (0.003, completed):** a $326,000 new wood-frame single-family house filed in 2012. Recent, large new construction rarely failed.
- **Typical (0.36, completed):** a $500 alteration filed in 2000 with no plan sets. It sits just above the 0.34 cutoff, so it is a false alarm; cases near the cutoff are close to a coin flip.
- **Most confident miss (0.01, did not finish):** a $3,600 alteration to a single-family home filed in 2013. Nothing on the application looks risky, and the model's strongest signal, filing year, says 2013 permits rarely fail.
- **Most confident false alarm (0.78, completed):** a 2005 alteration to a restaurant with a $0/$1 placeholder cost and no plan sets. Placeholder costs and missing plans usually go with unfinished projects, but not always.

The predictions tell the expected story: the model relies on filing year, cost and how complete the paperwork is.

### b. Wrong vs. right

```python
# --- Step 9b — how the four groups differ ---
order = ['caught failure', 'missed failure', 'false alarm', 'correct "completed"']
g = errs.groupby('result')
profile = pd.DataFrame({
    'permits': g.size(),
    'mean filing year': g['filing_year'].mean(),
    'median cost ($)': g['cost ($)'].median(),
    '% $0/$1 cost': g['cost_is_placeholder'].mean() * 100,
    'mean plan sets': g['plansets'].mean(),
    'median proposed stories': g['number_of_proposed_stories'].median(),
    '% site permit': g['site_permit'].apply(lambda s: (s == 'Y').mean() * 100),
    'mean predicted risk': g['predicted risk'].mean(),
}).loc[order]
profile.round(2)
```

```python
# --- Step 9b — share of failures caught (recall) and share of flags that are false alarms, by subgroup ---
errs['year band'] = pd.cut(errs['filing_year'], [1999, 2002, 2005, 2008, 2010, 2013],
                           labels=['2000–02', '2003–05', '2006–08', '2009–10', '2011–13'])
errs['cost band'] = pd.cut(errs['cost ($)'], [0, 5_000, 25_000, 100_000, 500_000, np.inf],
                           labels=['< $5k', '$5k–25k', '$25k–100k', '$100k–500k', '> $500k'])
errs['cost band'] = errs['cost band'].cat.add_categories('$0/$1 or missing').fillna('$0/$1 or missing')

def by_group(col):
    t = errs.groupby(col, observed=True).agg(
        permits=('result', 'size'),
        failure_rate=('actual', lambda s: (s == 'did not finish').mean()),
        failures_caught=('result', lambda s: (s == 'caught failure').sum() / s.isin(['caught failure', 'missed failure']).sum()),
        flags_that_are_false_alarms=('result', lambda s: (s == 'false alarm').sum() / max(s.isin(['caught failure', 'false alarm']).sum(), 1)))
    return t.round(3)

for col in ['year band', 'cost band', 'permit_type_definition']:
    display(by_group(col))
```

**Reading the comparison:**
- **Missed failures look like completed permits.** They were filed later than the caught failures (mean 2005 against 2003), cost more (median $50,000 against $30,000) and rarely have placeholder costs (3.7% against 9.6%), almost exactly the profile of the correctly predicted completions. Nothing on their application sets them apart.
- **False alarms look like caught failures:** older, cheaper and more often with placeholder costs. These are risky-looking applications that finished anyway.
- **Recent permits are the main blind spot.** The share of failures caught falls from 85% for permits filed 2000–02 to 43% for 2009–10 and 27% for 2011–13. Failure rates fell over time, and the model learned that recent permits are safe.
- **Large projects are harder too:** 50% of failures caught above $500,000 and 44% for new wood-frame construction, against 74% for alterations. These groups are small (762 and 602 test permits).
- **False alarms are spread evenly:** about half of the flags are false alarms in every group, as expected for a cutoff that flags 54% of permits.

### c. Acting on it

```python
# --- Step 9c — does the permit description (free text, not yet used) capture what the model is missing? ---
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
from permits_data import load_permits, resolved_permits

desc = resolved_permits(load_permits()).reset_index(drop=True)['description'].fillna('')
X_train_d, X_test_d = X_train.assign(description=desc.loc[X_train.index]), X_test.assign(description=desc.loc[X_test.index])

def with_text():
    return ColumnTransformer([('base', make_preprocessor(), list(X_train.columns)),
                              ('text', TfidfVectorizer(min_df=20, ngram_range=(1, 2), max_features=2000), 'description')])

rows = []
for name, prep, Xtr, Xte in [('without description', make_preprocessor(), X_train, X_test),
                             ('with description', with_text(), X_train_d, X_test_d)]:
    m = Pipeline([('prep', prep), ('model', LogisticRegression(max_iter=3000, C=0.1, class_weight='balanced'))]).fit(Xtr, y_train)
    pr = m.predict_proba(Xte)[:, 1]
    rows.append({'logistic regression': name, 'test ROC AUC': roc_auc_score(y_test, pr),
                 'recall on the forest\'s missed failures': (pr[(errs['result'] == 'missed failure').to_numpy()] >= 0.5).mean()})
pd.DataFrame(rows).set_index('logistic regression').round(3)
```

**What to do about it:**
1. **Add the missing information.** No model uses the free-text permit description. Adding it to logistic regression raises test AUC from 0.633 to 0.665, and catches 33% of the forest's missed failures instead of 27%. The description carries some of what the structured fields miss, so it is the first feature to add.
2. **Don't rely on the model for recent permits as it stands.** It catches only 27% of 2011–13 failures because it learned that later filings fail less. It should be retrained on recent resolved permits before use (Step 7).
3. **Treat large new construction with caution.** Fewer of its failures are caught, and there are too few of these permits to model separately.
4. **Accept that some failures can't be predicted from the application.** Many missed failures have an application that looks just like a completed one: the reasons a project is abandoned, such as financing or the owner's circumstances, are not on the form.

## Step 10 — Conclusions

### Which model best fits DBI's goal?
**The random forest.** DBI needs a ranked list of the riskiest new permits, and the forest:
1. has the highest test AUC (0.698), with every gap to the other models confirmed by the bootstrap,
2. gives the most accurate review list (65% of its riskiest 10% fail, against a 37% base rate),
3. has the highest accuracy and precision at the default cutoff.

**Trade-offs:**
- **Interpretability:** the forest can only be explained through feature importance. Logistic regression (odds ratios) or the depth-3 tree in `03` can be used alongside it to explain why a permit is flagged.
- **Later permits:** on 2011–2013 permits the forest's lead shrinks to a tie with the network (0.614), so the model should be retrained on recent data before use.

### Recommended decision policy
Use the forest to choose which permits get outreach, with a cutoff of **0.34** if a missed failure costs about twice a false alarm (Step 8). This costs about 24% less than contacting every permit, the best policy without a model. If staff capacity is limited, contact the riskiest permits up to that capacity instead: the riskiest 10% fail at 65%.

### What the errors show
The model misses failures whose applications look like completed permits, especially recent permits (only 27% of 2011–13 failures caught), because the application lacks the information that explains them (Step 9). Adding the permit description and retraining on recent permits are the next steps.

### Why the forest wins on this data
The data is tabular: a few dozen mixed numeric and categorical columns, many of them one-hot. Trees split directly on thresholds such as "filed before 2003" and on single categories, and averaging 200 trees removes most of a single tree's instability. Logistic regression can only fit straight-line effects, and the network has to approximate sharp jumps with smooth curves.

### Answer to the question
Application details predict whether a permit will be abandoned **better than chance but modestly**: AUC about 0.70 on a random split and about 0.61 on later permits. That is enough to prioritize outreach to high-risk permits, but not to deny or delay permits.
